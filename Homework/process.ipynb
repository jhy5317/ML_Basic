# 머신러닝 기법 적용 순서

## 1. sklearn 라이브러리 정리

### 1-1. model_selection
    train_test_split: train, test 데이터 분리
    KFold: K-Folding시 사용

### 1-2. metrics
    confusion_matrix: 혼동 행렬보기
    accuracy_score: 정확도
    precision_score: 정밀도
    recall_score: 재현율
    f1_score: f1스코어
    roc_auc_score: roc-auc 스코어
    precision_recall_curve: precision-recall 곡선 만들 때 사용

## 2. 데이터 확인 후 전처리
### 2-1 info(), head(), describe(), value_counts() 등 여러 방법을 통해서 적절한 데이터 처리법 정하기
    이상치, 결측치 등 처리하기

### 2-2 데이터 셋을 X_features(학습데이터), y_label(정답지)로 나눠서 저장
    ex) X_features = data.iloc / y_label = target.iloc~ 등 여러 기법을 사용해서 문제와 정답 나누기

## 3. Train, Validation, Test 셋 나누기
### 3-1 X_train, X_val, X_test, y_train, y_val, y_test 나누기 (train_test_split 사용 -> from sklearn.model_selection)
    validation은 XGBoost, LightGBM에서 사용
    from sklearn.model_selection import train_test_split
    X_train, X_test, y_train, y_test = train_test_split(문제지, 정답지, 테스트사이즈, random_state)
    문제지: 보통 X, 정답지 : 보통 y, 테스트사이즈: test_size = 0.x, random_state = 아무숫자

    만약 부스트 모델 사용할 경우
    X_train, y_train 을 한번더 train_test_split으로
    X_tr, X_val, y_tr, y_val로 나눠서사용
