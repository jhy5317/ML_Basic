```python
# 7 Clustering
## K-Means

from sklearn.preprocessing import scale
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

iris = load_iris()

df = pd.DataFrame(data = iris.data, columns = iris.feature_names)
df.head()
```

```python
kmeans = KMeans(n_clusters = 3, init = "k-means++", max_iter = 300, random_state = 42)

kmeans.fit(df);

print(kmeans.labels_)
```

```python
# 원래 정답과 cluster에서 분류한 결과랑 비교하기

df["target"] = iris.target
df["cluster"] = kmeans.labels_

res = df.groupby(["target", "cluster"])["sepal length (cm)"].count()
print(res)
```

```python
# PCA로 차원축소

from sklearn.decomposition import PCA

pca = PCA(n_components = 2)
df_pca = pca.fit_transform(iris.data)

df["pca_1"] = df_pca[:, 0]
df["pca_2"] = df_pca[:, 1]

df.head()
```

```python
marker0 = df[df["cluster"] == 0].index
marker1 = df[df["cluster"] == 1].index
marker2 = df[df["cluster"] == 2].index

plt.scatter(x = df.loc[marker0, "pca_1"], y = df.loc[marker0, "pca_2"], marker = "o")
plt.scatter(x = df.loc[marker1, "pca_1"], y = df.loc[marker1, "pca_2"], marker = "s")
plt.scatter(x = df.loc[marker2, "pca_1"], y = df.loc[marker2, "pca_2"], marker = "^")

plt.xlabel("PCA 1")
plt.ylabel("PCA 2")

plt.show()
```

```python
# Clustering Algorithm
## 군집 중심 찍기

import numpy as np
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs # 클러스터링 하기 좋은 임의의 데이터 생성 함수

X, y = make_blobs(n_samples = 200, n_features = 2, centers = 3, cluster_std = 1.5, random_state = 23)
# cluster_std: 표준편차가 작을수록 많이 모임 -> 추후에 실루엣 점수와 연관

print(X.shape, y.shape)
```

```python
unique, counts = np.unique(y, return_counts = True)
print(unique, counts)
```

```python
import pandas as pd

df_cls = pd.DataFrame(data = X, columns = ["ftr1", "ftr2"])
df_cls["target"] = y
df_cls.head()
```

```python
target_list = np.unique(y)

markers = ["o", "s", "^", "P", "D", "H", "x"]

for target in target_list:
    target_cluster = df_cls[df_cls["target"] == target]
    plt.scatter(x = target_cluster["ftr1"], y = target_cluster["ftr2"], edgecolors = "k", marker = markers[target])

plt.show()
```

```python
kmeans = KMeans(n_clusters = 3, init = "k-means++", max_iter = 200, random_state = 42)

cluster_labels = kmeans.fit_predict(X)
df_cls["kmeans_label"] = cluster_labels

# 센터 찾기
centers = kmeans.cluster_centers_
unique_labels = np.unique(cluster_labels)

markers = ["o", "s", "^", "P", "D", "H", "x"]

for label in unique_labels:
    label_cluster = df_cls[df_cls["kmeans_label"] == label]
    center = centers[label]
    plt.scatter(x = label_cluster["ftr1"], y = label_cluster["ftr2"], edgecolors = "k", marker = markers[label])
    plt.scatter(x = center[0], y = center[1], s = 200, color = "white", alpha = 0.9, edgecolors = "k", marker = markers[label])
    plt.scatter(x = center[0], y = center[1], s = 70, color = "k", edgecolors = "k", marker = "$%d$" % label)

plt.show()

# 멀리 동 떨어진 데이터가 많이 있으면 군집 갯수에 대한 고찰을 해봐야함.
```

```python
print(df_cls.groupby("target")["kmeans_label"].value_counts())
```

```python
import numpy as np

from sklearn.datasets import make_blobs
from sklearn.cluster import MeanShift

X, y = make_blobs(n_samples = 200, n_features = 2, centers = 3, cluster_std = 0.7, random_state = 42)
# cluster_std: 표준편차 -> 클 수록 많이 나눠짐

meanshift = MeanShift(bandwidth = 1.0) # 적을수록 많이 나눠짐 -> 같은 클러스터라고 인식하는 범위라고 생각하면 편함
cluster_labels = meanshift.fit_predict(X)

print(np.unique(cluster_labels))
```

```python
from sklearn.cluster import estimate_bandwidth

bandwidth = estimate_bandwidth(X)
print(round(bandwidth, 3))
```

```python
import pandas as pd

df = pd.DataFrame(data = X, columns = ["ftr1", "ftr2"])
df["target"] = y

best_bandwidth = estimate_bandwidth(X)

meanshift = MeanShift(bandwidth = best_bandwidth)

cluster_labels = meanshift.fit_predict(X)

print(np.unique(cluster_labels))
```

```python
import matplotlib.pyplot as plt

df["meanshift_label"] = cluster_labels
centers = meanshift.cluster_centers_
unique_labels = np.unique(cluster_labels)

markers = ["o", "s", "^", "x", "*"]

for label in unique_labels:
    label_cluster = df[df["meanshift_label"] == label]
    center = centers[label]

    plt.scatter(x = label_cluster["ftr1"], y = label_cluster["ftr2"], edgecolors = "k", marker = markers[label])
    plt.scatter(x = center[0], y = center[1], s = 200, color = "white", alpha = 0.9, edgecolors = "k", marker = markers[label])
    plt.scatter(x = center[0], y = center[1], s = 70, color = "k", edgecolors = "k", marker = "$%d$" % label)

plt.show()
```

```python
print(df.groupby("target")["meanshift_label"].value_counts())
```
